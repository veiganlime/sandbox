```python
from datetime import datetime
from dateutil.relativedelta import relativedelta
import pandas as pd
```

```python
today = datetime.now()
last_extraction = "2015-12-01"
start_extraction = ""
set_extraction = ""

last_extraction = pd.to_datetime(last_extraction)
today = pd.to_datetime(today)

print(today)
print(last_extraction)
print(start_extraction)
```

```python
if today.year == last_extraction.year:
    print("The years are the same")
    if today.month > last_extraction.month:
        print("The last Extraction month is smaller than today's month!")
        month_diff = today.month - last_extraction.month
        set_extraction = last_extraction + relativedelta(months=1)
        print("Month difference:", month_diff)
        print("Set Extraction date:", set_extraction)

elif today.year > last_extraction.year:
    print("The last extraction year is smaller than today's year")
    year_diff = today.year - last_extraction.year
    print("Year difference:", year_diff)
    
    month_diff = (year_diff * 12) + today.month - last_extraction.month
    print("Month difference:", month_diff)
    set_extraction = last_extraction + relativedelta(months=1)
    print("Set Extraction date:", set_extraction)

else:
    print("Value Error! - input date isnt correct")
```
